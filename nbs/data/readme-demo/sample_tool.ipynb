```
#| export
def double(x):
    "Return x multiplied by 2."
    return x * 2
```

`double` is a tiny, exported helper used by the demo notebook and any downstream readers of `sample_tool`.
Keeping it minimal makes the behavior easy to reuse and verify.

```
# focused test
assert double(4) == 8
```

```
# visible example
print(double(3))
```
